### Bài 1: Tạo quần thể mô phỏng

Bạn hãy tạo một mẫu gồm 10000 phần tử mô phỏng chiều cao của nam thanh niên có giá trị từ 120cm - 200cm. Bạn hãy lưu kết quả vào biến **POP**.

Tính:
a. Chiều cao trung bình (kỳ vọng) của quần thể và độ lệch chuẩn về chiều cao của quần thể.

b. Tính tỷ lệ người cao trong quần thể, biết rằng thanh niên có chiều cao từ 180 trở lên được gọi là cao.

c. Vẽ histogram về chiều cao của quần thể. Theo bạn quần thể có phân phối chuẩn hay không?

d. Thử vẽ histogram và đồ thị hàm mật độ của phân phối chuẩn sử dụng tham số loc và scale bằng với kỳ vọng và độ lệch chuẩn của quần thể.

```python
import scipy.stats as stats
import matplotlib.pyplot as plt
import numpy as np
POP = np.random.uniform(120, 200, 10000)

# a. Chiều cao trung bình (kỳ vọng) của quần thể và độ lệch chuẩn về chiều cao của quần thể.
mean_height = np.mean(POP)
std_height = np.std(POP)
print(f"Chiều cao trung bình: {mean_height:.2f} cm")
print(f"Độ lệch chuẩn: {std_height:.2f} cm")

# b. Tính tỷ lệ người cao trong quần thể
tall_people = POP[POP >= 180]
ratio_tall = len(tall_people) / len(POP)
print(f"Tỷ lệ người cao: {ratio_tall:.2%}")

# c. Vẽ histogram về chiều cao của quần thể
plt.figure(figsize=(10, 6))
plt.hist(POP, bins=30, density=True, alpha=0.6, color='g', label='Population')
plt.title('Histogram of Population Heights')
plt.xlabel('Height (cm)')
plt.ylabel('Density')
plt.legend()
plt.show()

# d. Vẽ histogram và đồ thị hàm mật độ của phân phối chuẩn
# Tạo phân phối chuẩn với kỳ vọng và độ lệch chuẩn của quần thể
x = np.linspace(120, 200, 1000)
normal_dist = stats.norm.pdf(x, loc=mean_height, scale=std_height)

# Vẽ histogram và hàm mật độ
plt.figure(figsize=(10, 6))
plt.hist(POP, bins=30, density=True, alpha=0.6, color='g', label='Population')
plt.plot(x, normal_dist, 'r-', lw=2, label='Normal Distribution')
plt.title('Histogram and PDF of Normal Distribution')
plt.xlabel('Height (cm)')
plt.ylabel('Density')
plt.legend()
plt.show()
```

### Bài 7:  Bài Tập Ước Lượng

Khảo sát cân nặng (kg) của gà khi xuất chuồng, người ta cân một số con và có kết quả như sau:

> 2.1, 1.8, 2.0, 2.3, 1.7, 1.5, 2.0, 2.2, 1.8

Giả sử cân nặng của gà là biến ngẫu nhiên có phân phối chuẩn. Hãy lập khoảng tin cậy cho cân nặng trung bình của  khi xuất chuồng với độ tin cậy là 95%. Trong 2 trường hợp:

a, Biết $\sigma = 0.3$

b, Không biết $\sigma$


```python
# import matplotlib.pyplot as plt
# import numpy as np
# from scipy import stats

# # Data for problem 7
# weights = np.array([2.1, 1.8, 2.0, 2.3, 1.7, 1.5, 2.0, 2.2, 1.8])

# # a. Known population standard deviation (sigma = 0.3)
# sigma = 0.3
# n = len(weights)
# mean_weight = np.mean(weights)
# z = stats.norm.ppf(0.975)  # 95% confidence interval
# margin_of_error = z * (sigma / np.sqrt(n))
# confidence_interval_a = (mean_weight - margin_of_error, mean_weight + margin_of_error)
# print(f"a. 95% Confidence Interval (known sigma): {confidence_interval_a}")


# # b. Unknown population standard deviation
# s = np.std(weights, ddof=1)  # Sample standard deviation
# t = stats.t.ppf(0.975, n - 1)  # t-distribution critical value
# margin_of_error = t * (s / np.sqrt(n))
# confidence_interval_b = (mean_weight - margin_of_error, mean_weight + margin_of_error)
# print(f"b. 95% Confidence Interval (unknown sigma): {confidence_interval_b}")


import numpy as np
from scipy import stats

# Dữ liệu cân nặng
data = np.array([2.1, 1.8, 2.0, 2.3, 1.7, 1.5, 2.0, 2.2, 1.8])
n = len(data)
mean = np.mean(data)
std_sample = np.std(data, ddof=1)  # ddof=1 để tính độ lệch chuẩn mẫu
confidence = 0.95
alpha = 1 - confidence

# Trường hợp a: Biết sigma
sigma = 0.3
z = stats.norm.ppf(1 - alpha/2)
ci_known_sigma = (
    mean - z * sigma / np.sqrt(n),
    mean + z * sigma / np.sqrt(n)
)

# Trường hợp b: Không biết sigma
t = stats.t.ppf(1 - alpha/2, df=n-1)
ci_unknown_sigma = (
    mean - t * std_sample / np.sqrt(n),
    mean + t * std_sample / np.sqrt(n)
)

# In kết quả
print("Trường hợp a - Biết σ = 0.3:")
print(f"Khoảng tin cậy 95%: {ci_known_sigma}")

print("\nTrường hợp b - Không biết σ:")
print(f"Độ lệch chuẩn mẫu: {std_sample:.3f}")
print(f"Khoảng tin cậy 95%: {ci_unknown_sigma}")
```

### Bài 8

Bộ dữ liệu `SAT-lard.csv` chứa điểm thi SAT của 1000 học sinh.

a. Giả sử bộ dữ liệu là kết quả của quá trình lấy mẫu ngẫu nhiên từ các học sinh. Bạn hãy xây dựng khoảng tin cậy cho trung bình điểm thi của tất cả các sinh viên với độ tin cậy 99% (`Đáp án: 105.2 ± 3.10`)

e. Giả sử bộ dữ liệu là toàn bộ kết quả điểm thi SAT của học sinh một trường phố thông. Sử dụng mẫu là điểm thi của 36 học sinh đầu tiên trong bộ dữ liệu, bạn hãy xây dựng khoảng tin cậy **95%** cho điểm SAT trung bình của trường phổ thông đấy. Cho biết khoảng ước lượng bạn xây dựng có đúng không, dựa trên giá trị đã tính ở câu trên.

```python
import pandas as pd
import numpy as np
from scipy import stats

# Đọc dữ liệu
df = pd.read_csv('./SAT-lard.csv')

# Giả sử cột chứa điểm thi SAT là 'SAT_score'
# Nếu cột khác tên, bạn cần thay đổi 'SAT_score' thành tên đúng của cột.
sat_scores = df['SAT Score']

# a. Khoảng tin cậy 99% cho trung bình điểm thi SAT
sample_mean = np.mean(sat_scores)
sample_std = np.std(sat_scores, ddof=1)  # Độ lệch chuẩn mẫu
sample_size = len(sat_scores)

# Xác định mức ý nghĩa
alpha = 0.01  # 99% độ tin cậy => alpha = 0.01

# Tính giá trị t cho khoảng tin cậy 99%
t_critical = stats.t.ppf(1 - alpha / 2, df=sample_size - 1)

# Tính khoảng tin cậy
margin_of_error = t_critical * (sample_std / np.sqrt(sample_size))
confidence_interval = (sample_mean - margin_of_error, sample_mean + margin_of_error)

# In kết quả câu a
print(f"a. Khoảng tin cậy 99%: {sample_mean:.2f} ± {margin_of_error:.2f}")
print(f"Khoảng tin cậy 99%: ({confidence_interval[0]:.2f}, {confidence_interval[1]:.2f})")

# b. Khoảng tin cậy 95% cho trung bình điểm thi SAT
alpha = 0.05  # 95% độ tin cậy => alpha = 0.05

# Tính giá trị t cho khoảng tin cậy 95%
t_critical_95 = stats.t.ppf(1 - alpha / 2, df=sample_size - 1)

# Tính khoảng tin cậy
margin_of_error_95 = t_critical_95 * (sample_std / np.sqrt(sample_size))
confidence_interval_95 = (sample_mean - margin_of_error_95, sample_mean + margin_of_error_95)

# In kết quả câu b
print(f"b. Khoảng tin cậy 95%: {sample_mean:.2f} ± {margin_of_error_95:.2f}")
print(f"Khoảng tin cậy 95%: ({confidence_interval_95[0]:.2f}, {confidence_interval_95[1]:.2f})")

# c. Tính điểm SAT trung bình của toàn bộ trường phổ thông
mu = np.mean(sat_scores)
print(f"c. Điểm SAT trung bình của toàn bộ trường phổ thông: μ = {mu:.2f}")

# d. Khoảng tin cậy 99% cho điểm SAT trung bình của trường phổ thông từ mẫu 36 học sinh đầu tiên
sample_36 = sat_scores[:36]
sample_mean_36 = np.mean(sample_36)
sample_std_36 = np.std(sample_36, ddof=1)
sample_size_36 = len(sample_36)

# Tính giá trị t cho khoảng tin cậy 99% với mẫu 36 học sinh
t_critical_36 = stats.t.ppf(1 - alpha / 2, df=sample_size_36 - 1)

# Tính khoảng tin cậy
margin_of_error_36 = t_critical_36 * (sample_std_36 / np.sqrt(sample_size_36))
confidence_interval_36 = (sample_mean_36 - margin_of_error_36, sample_mean_36 + margin_of_error_36)

# In kết quả câu d
print(f"d. Khoảng tin cậy 99% từ mẫu 36 học sinh: ({confidence_interval_36[0]:.2f}, {confidence_interval_36[1]:.2f})")

# e. Khoảng tin cậy 95% cho điểm SAT trung bình của trường phổ thông từ mẫu 36 học sinh
alpha_95 = 0.05  # 95% độ tin cậy => alpha = 0.05

# Tính giá trị t cho khoảng tin cậy 95% với mẫu 36 học sinh
t_critical_36_95 = stats.t.ppf(1 - alpha_95 / 2, df=sample_size_36 - 1)

# Tính khoảng tin cậy
margin_of_error_36_95 = t_critical_36_95 * (sample_std_36 / np.sqrt(sample_size_36))
confidence_interval_36_95 = (sample_mean_36 - margin_of_error_36_95, sample_mean_36 + margin_of_error_36_95)

# In kết quả câu e
print(f"e. Khoảng tin cậy 95% từ mẫu 36 học sinh: ({confidence_interval_36_95[0]:.2f}, {confidence_interval_36_95[1]:.2f})")
```

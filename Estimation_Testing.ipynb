# Phần 1: Ước lượng

### 📋 BẢNG TỔNG HỢP CÔNG THỨC ƯỚC LƯỢNG & TRƯỜNG HỢP SỬ DỤNG Z/T

| Trường hợp                                                 | Số lượng mẫu       | Biết σ?    | Dùng z/t     | Công thức khoảng tin cậy |
|------------------------------------------------------------|--------------------|------------|--------------|----------------------------|
| Ước lượng trung bình (biết σ, phân phối chuẩn hoặc n ≥ 30) | Lớn (n ≥ 30) hoặc chuẩn | Có         | z            | $\bar{x} \pm z_{\frac{\alpha}{2}} \cdot \frac{\sigma}{\sqrt{n}} $ |
| Ước lượng trung bình (không biết σ, n nhỏ)                 | Nhỏ (n < 30)       | Không       | t (Student)  | $ \bar{x} \pm t_{\frac{\alpha}{2},n-1} \cdot \frac{s}{\sqrt{n}} $ |
| Ước lượng trung bình (không biết σ, n lớn)                 | Lớn (n ≥ 30)       | Không       | xấp xỉ z (dùng t cũng được)     | $ \bar{x} \pm z_{\frac{\alpha}{2}} \cdot \frac{s}{\sqrt{n}} $ |
| Ước lượng tỷ lệ (proportion)                              | np ≥ 5, n(1-p) ≥ 5 | -          | z            | $\hat{p} \pm z_{\frac{\alpha}{2}} \cdot \sqrt{ \frac{ \hat{p}(1 - \hat{p}) }{n} } $ |

### 🎯 Công thức tính trị tới hạn (Critical Value)

| Loại kiểm định      | Dạng kiểm định | Công thức trị tới hạn (Python)                                | Ý nghĩa |
|---------------------|----------------|----------------------------------------------------------------|---------|
| **Z (biết σ)**      | 2 phía         | `z_crit = norm.ppf(1 - α/2)`                                   | $z_{\alpha/2} $ |
|                     
| **T (không biết σ)**| 2 phía         | `t_crit = t.ppf(1 - α/2, df=n-1)`                              | $ t_{\alpha/2, n-1} $ |
|                 


```python
import pandas as pd
import numpy as np
import scipy.stats as stats
```

### Bài 1: Biết $\sigma$, $n\ge30$. Dùng z
Một nhà máy sản xuất pin. Mẫu gồm 50 viên pin có tuổi thọ trung bình là 82.3 giờ. Độ lệch chuẩn biết trước là 8 giờ. Tính khoảng tin cậy 95% cho tuổi thọ pin trung bình.

Theo định lý giới hạn trung tâm, với mẫu lớn (n ≥ 30), phân phối của trung bình mẫu $\bar{x}$ sẽ gần như chuẩn, dù tổng thể không chuẩn.

Do đã biết σ (độ lệch chuẩn tổng thể), ta dùng công thức cho khoảng tin cậy với z-score. Đây là công thức chính xác nhất khi biết σ.

Tóm tắt: \
$n = 50 ≥ 30$\
$\bar{x} = 82.3$\
$\sigma = 8$  
$\alpha = 1 - 0.95 = 0.05 $

Khoảng tin cậy:
$$\bigg(\bar{x} - z_{\frac{\alpha}{2}}\frac{\sigma}{\sqrt{n}}, \bar{x}+z_{\frac{\alpha}{2}}\frac{\sigma}{\sqrt{n}}\bigg)$$

```python
# Dữ liệu
x_bar = 82.3
sigma = 8
n = 50
alpha = 0.05
z = stats.norm.ppf(1 - alpha/2)

# Tính SE và CI
se = sigma / np.sqrt(n)
ci = (x_bar - z * se, x_bar + z * se)

print(f"Khoảng tin cậy 95%: {ci}")
```

### Bài 2: Không biết $\sigma$, $n\le30$. Dùng t
Mẫu gồm 10 người có cân nặng như sau: `[60, 62, 58, 65, 61, 63, 64, 59, 60, 62]`. Tính khoảng tin cậy 95% cho trung bình cân nặng. Giả định tổng thể có phân phối chuẩn

Vì cỡ mẫu nhỏ (n = 10) và không biết σ, có phân phối chuẩn nên dùng phân phối t

Tóm tắt: \
$n = 10 \le 30$\
$\bar{x} = 61.4$\
$s=2.22$\
$\alpha = 1 - 0.95 = 0.05 $

Khoảng tin cậy:

$$\bigg(\bar{x} - t_{\frac{\alpha}{2}, n-1}\frac{s}{\sqrt{n}},\bar{x}+t_{\frac{\alpha}{2}, n-1}\frac{s}{\sqrt{n}}\bigg)$$

```python
data = [60, 62, 58, 65, 61, 63, 64, 59, 60, 62]
n = len(data)
x_bar = np.mean(data)
s = np.std(data, ddof=1)

t_score = stats.t.ppf(1 - alpha/2, df=n-1)
se = s / np.sqrt(n)
ci = (x_bar - t_score * se, x_bar + t_score * se)

print("Trung bình mẫu: ", x_bar)
print("Độ lệch chuẩn mẫu: ", s)
print(f"Khoảng tin cậy 95%: {ci}")
```

### Bài 3: Không biết $\sigma$, $n\ge30$. Dùng t
Mẫu gồm 40 người có chiều cao trung bình là 170.2 cm, độ lệch chuẩn mẫu là 6.5. Tính khoảng tin cậy 99% cho chiều cao trung bình.



Dù không biết σ, nhưng với mẫu lớn (n ≥ 30), theo CLT, $\bar{x}$ xấp xỉ phân phối chuẩn -> dùng phân phối t

Tóm tắt: \
$n = 40 \ge 30$\
$\bar{x} = 170.2$\
$s=6.5$\
$\alpha = 1 - 0.99 = 0.01 $

Khoảng tin cậy:

$$\bigg(\bar{x} - t_{\frac{\alpha}{2}, n-1}\frac{s}{\sqrt{n}},\bar{x}+t_{\frac{\alpha}{2}, n-1}\frac{s}{\sqrt{n}}\bigg)$$

```python
x_bar = 170.2
s = 6.5
n = 40
alpha = 0.01
t_score = stats.t.ppf(1 - alpha/2, df=n-1)
se = s / np.sqrt(n)
ci = (x_bar - t_score * se, x_bar + t_score * se)
print(f"CI 99%: {ci}")
```

## Bài 4: Ước lượng tỷ lệ. Dùng z
Khảo sát 200 người, có 142 người thích sản phẩm. Ước lượng tỉ lệ người thích sản phẩm trong tổng thể với độ tin cậy 95%.

Tóm tắt:\
$n = 200 \ge 30$\
$\hat{p} = 170.2$\
$\alpha = 1 - 0.95 = 0.05 $

- Thích hay không thích là độc lập với nhau nên tỷ lệ người thích tuân theo phân phối nhị thức
- Ta có
    - $\hat{p}= 0.71$, $n\hat{p} = 142 \geq 5$
    - $n(1-\hat{p}) = 58 \geq 5$
    
=> Vậy phân phối nhị thức của tỷ lệ mẫu có thể xấp xỉ bằng phân phối chuẩn. Ta có công thức thống kê kiểm định tính như sau:
$$z=\frac{\hat{p}-p}{\sqrt{\frac{pq}{n}}}$$

Khoảng tin cậy:
$$\left( \hat{p} - z_{\frac{\alpha}{2}} \cdot \sqrt{\frac{\hat{p}(1 - \hat{p})}{n}}, \quad \hat{p} + z_{\frac{\alpha}{2}} \cdot \sqrt{\frac{\hat{p}(1 - \hat{p})}{n}} \right)$$


```python
from scipy.stats import norm

n = 200
x = 142
p_hat = x / n
alpha = 0.05
z = norm.ppf(1 - alpha/2)
se = np.sqrt(p_hat * (1 - p_hat) / n)
ci = (p_hat - z*se, p_hat + z*se)
print("Trung bình tỷ lệ người thích: ",p_hat )

print(f"CI 95% cho tỉ lệ: {ci}")
```

## Bài 5: Trường hợp đặc biệt. Dùng bootstrap
Khảo sát thời gian ngủ (tiếng) của 10 người dùng smartphone trước khi ngủ: `data = [5.5, 6.2, 4.8, 6.0, 5.7, 5.1, 6.5, 4.9, 5.3, 5.6]`. Dựng khoảng tin cậy 95% cho thời gian ngủ trung bình.



Khi không biết phân phối và không thỏa mãn điều kiện dùng t, ta cần dùng bootstrap — một phương pháp thống kê phi tham số, không cần giả thiết về phân phối.


```python
# Cách giải
# Tạo hàng nghìn mẫu bootstrap (lấy lại từ mẫu gốc).

# Tính trung bình cho mỗi mẫu.

# Lấy 2.5% và 97.5% làm khoảng tin cậy.

import numpy as np

data = [5.5, 6.2, 4.8, 6.0, 5.7, 5.1, 6.5, 4.9, 5.3, 5.6]
means = []

np.random.seed(0)
for _ in range(10000):
    sample = np.random.choice(data, size=len(data), replace=True)
    means.append(np.mean(sample))

ci = np.percentile(means, [2.5, 97.5])
print("Khoảng tin cậy bootstrap 95%:", ci)
```

# Phần 2: Kiểm định

## Tổng hợp công thức:

### 🧪 Công thức tính trị kiểm định (Test Statistic)

| Tên kiểm định            | Công thức trị kiểm định                                               | Ý nghĩa |
|--------------------------|----------------------------------------------------------------------|--------|
| **Z - Trung bình (biết σ)** | $ z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}} $                    | So sánh khi biết độ lệch chuẩn tổng thể |
| **T - Trung bình (không biết σ)** | $ t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}} $ | Khi không biết σ, dùng độ lệch chuẩn mẫu |
| **Z - Tỷ lệ**               | $ z = \frac{\hat{p} - p_0}{\sqrt{p_0(1 - p_0)/n}} $                    | Kiểm định tỷ lệ |

### 📊 Cách tính p-value theo loại kiểm định

| Loại kiểm định | Dạng kiểm định | Công thức tính p-value (Python)                                   |
|----------------|----------------|---------------------------------------------------------------------|
| **Z (biết σ)** | 2 phía         | `2 * (1 - norm.cdf(abs(z)))`                                       |
|                | Phía trái      | `norm.cdf(z)`                                                      |
|                | Phía phải      | `1 - norm.cdf(z)`                                                  |
| **T (không biết σ)** | 2 phía         | `2 * (1 - t.cdf(abs(t), df=n-1))`                            |
|                | Phía trái      | `t.cdf(t, df=n-1)`                                                 |
|                | Phía phải      | `1 - t.cdf(t, df=n-1)`                                             |

### 🚫 Miền bác bỏ (Vùng từ chối H₀)

| Loại kiểm định      | Dạng kiểm định | Miền bác bỏ (với mức ý nghĩa α)                     |
|---------------------|----------------|-----------------------------------------------------|
| **Z (biết σ)**      | 2 phía         | $|z| > z_{\alpha/2}$                              |
|                     | Phía trái      | $z < -z_{\alpha}$                                 |
|                     | Phía phải      | $z > z_{\alpha}$                                  |
| **T (không biết σ)**| 2 phía         | $|t| > t_{\alpha/2, n-1}$                         |
|                     | Phía trái      | $t < -t_{\alpha, n-1}$                            |
|                     | Phía phải      | $t > t_{\alpha, n-1}$                             |

### 🎯 Công thức tính trị tới hạn (Critical Value)

| Loại kiểm định      | Dạng kiểm định | Công thức trị tới hạn (Python)                                | Ý nghĩa |
|---------------------|----------------|----------------------------------------------------------------|---------|
| **Z (biết σ)**      | 2 phía         | `z_crit = norm.ppf(1 - α/2)`                                   | $z_{\alpha/2} $ |
|                     | Phía trái      | `z_crit = norm.ppf(α)`                                         | $ z_{\alpha} $ (âm) |
|                     | Phía phải      | `z_crit = norm.ppf(1 - α)`                                     | $ z_{\alpha} $ |
| **T (không biết σ)**| 2 phía         | `t_crit = t.ppf(1 - α/2, df=n-1)`                              | $ t_{\alpha/2, n-1} $ |
|                     | Phía trái      | `t_crit = t.ppf(α, df=n-1)`                                    | $ t_{\alpha, n-1} $ (âm) |
|                     | Phía phải      | `t_crit = t.ppf(1 - α, df=n-1)`                                | $ t_{\alpha, n-1} $ |  




## Bài 1: Kiểm định phía trái
Một cửa hàng thực phẩm nhận thấy thời gian vừa qua trung bình một khách hàng mua 25 ngàn đồng thực phẩm trong ngày. Nay cửa hàng chọn ngẫu nhiên 15 khách hàng thấy trung bình một khách hàng mua 24 ngàn đồng trong ngày và phương sai mẫu là $s^2$ =  (2 ngàn đồng $)^2$

   Với mức ý nghĩa là 5%, kiểm định xem có phải sức mua của khách hàng hiện nay thực sự giảm sút hay không. Biết rằng sức mua của khách hàng có phân phối chuẩn.

Tóm tắt:
- $\mu_0 = 25$
- n = 15
- $\alpha = 0.05$
- $\bar{x} = 24$
- $s^{2} = 4$
- $s = 2$

**B1** Xây dựng giả thuyết kiểm định

$$H_o: \mu = 25$$

$$H_a: \mu < 25$$

**B2** Mức ý nghĩa của kiểm định $\alpha = 0.05$

**B3** Tính giá trị kiểm định từ mẫu
$$t_0= \frac{\bar{x} - \mu_0}{\frac{s}{\sqrt{n}}} \approx -1.9365$$

**B4** Xác định `p-value` hoặc `miền bác  bỏ` (critical region) của kiểm định

- Trường hợp xác định miền bác bỏ

$$\{ t: t < -t_{\alpha} \}$$

$$\{ t: t < -1.7613 \}$$

- Trường hợp xác định p_value
$$ p\_value = 0.0366 < 0.05 $$

**B5** Đưa ra giả thuyết vô hiệu. Rút ra kết luận chung
- Bác bỏ giả thuyết $H_0$ vì trị kiểm định thuộc miền bác bỏ (p_value < mức ý nghĩa)
- Điều này có nghĩa là sức mua hiện nay đã thực sự sụt giảm

```python
mu = 25
tb = 24
dlc = 2
n = 15

trikiemdinh = ((tb - mu) / (dlc / np.sqrt(n))).round(4)
print('- Trị kiểm định: ', trikiemdinh)

print('- Giá trị tới hạn: ', stats.t.ppf(0.05, n-1).round(4))

print('- p_value: ', stats.t.cdf(trikiemdinh, df=n-1).round(4))
```

## Bài 2: Kiểm định phía phải
Trong điều kiện chăn nuôi bình thường, lượng sữa trung bình của 1 con bò là 14 kg/ngày. Có ý kiến cho rằng sau khi cải thiện điều kiện chăn nuôi, lượng sữa trung bình của bò đã tăng lên.

Để kiểm tra điều này, người ta chọn ngẫu nhiên 25 con bò, đo lượng sữa mỗi ngày và thu được:

Trung bình mẫu: $x =15.2$ kg/ngày

Độ lệch chuẩn mẫu: $s=2.4$ kg

Với mức ý nghĩa $α=0.05$, hãy kiểm định giả thuyết trên.
Biết rằng lượng sữa bò có phân phối chuẩn.

Tóm tắt:
- $\mu_0 = 14$
- n = 25
- $\alpha = 0.05$
- $\bar{x} = 12.5$
- $s = 2.5$

**B1** Xây dựng giả thuyết kiểm định

$$H_o: \mu = 14$$

$$H_a: \mu > 14$$

**B2** Mức ý nghĩa của kiểm định $\alpha = 0.05$

**B3** Tính giá trị kiểm định từ mẫu
$$t_0= \frac{\bar{x} - \mu_0}{\frac{s}{\sqrt{n}}} \approx -3.000$$

**B4** Xác định `p-value` hoặc `miền bác  bỏ` (critical region) của kiểm định

- Trường hợp xác định miền bác bỏ

$$\{ t: t > t_{\alpha} \}$$

$$\{ t: t > 1.7109 \}$$

- Trường hợp xác định p_value
$$ p\_value = 0.99 > 0.05 $$

**B5** Đưa ra giả thuyết vô hiệu. Rút ra kết luận chung
- Không bác bỏ giả thuyết $H_0$ vì trị kiểm định không thuộc miền bác bỏ (p_value > mức ý nghĩa)
- Không đủ bằng chứng thống kê để nói rằng lượng sữa bò đã tăng lên.

```python
from scipy.stats import t
import numpy as np

# Dữ liệu
mu_0 = 14
x_bar = 12.5
s = 2.5
n = 25
alpha = 0.05

# Tính t0
t_stat = (x_bar - mu_0) / (s / np.sqrt(n))

# Trị tới hạn
t_crit = t.ppf(1 - alpha, df=n-1)

# P-value
p_value = 1 - t.cdf(t_stat, df=n-1)

print(f"Trị kiểm định: {t_stat:.4f}")
print(f"Trị tới hạn: {t_crit:.4f}")
print(f"p-value = {p_value:.4f}")
```

Nay để xác định lại trọng lượng ấy, người ta chọn ngẫu nhiên 100 thanh niên đo trọng lượng trung bình là 50 kg và phương sai mẫu $s^2 = (10 kg)^2$.
```python
mu = 48
tb = 50
dlc = 10
n = 100

trikiemdinh = ((tb - mu) / (dlc / np.sqrt(n))).round(4)
print('- Trị kiểm định: ', trikiemdinh)

# Tính giá trị tới hạn để xác định miền bác bỏ
alpha = 0.01
tritoihan = t.ppf(1 - alpha/2, df=n-1)
print('- Trị tới hạn: ', tritoihan.round(4))

print('- p_value: ', t.cdf(trikiemdinh, df=n-1).round(4))
```

## Bài 4: Kiểm định tỷ lệ 2 phía

Tỷ lệ phế phẩm của một nhà máy trước đây là 5%. Năm nay nhà máy áp dụng một biện pháp kỹ thuật mới. Để nghiên cứu tác dụng của biện pháp kỹ thuật mới, người ta lấy một mẫu gồm 800 sản phẩm để kiểm tra và thấy có 24 phế phẩm.

a. Với α = 0.01. Hãy cho kết luận về biện pháp kỹ thuật mới này?

b. Nếu nhà máy báo cáo tỷ lệ phế phẩm sau khi áp dụng biện pháp kỹ thuật mới là
2% thì có chấp nhận được không? (α = 0.01).

**B4** Xác định `p-value` hoặc `miền bác  bỏ` (critical region) của kiểm định

**B5** Đưa ra giả thuyết vô hiệu. Rút ra kết luận chung
- Bác bỏ giả thuyết $H_0$ vì trị kiểm định thuộc miền bác bỏ
- Điều này có nghĩa là biện pháp kỹ thuật mới thay đổi tỷ lệ phế phẩm

```python
# Tính giá trị tới hạn để xác định miền bác bỏ
alpha = 0.01
tritoihan = stats.norm.ppf(1 - alpha/2).round(4)
print('- Trị tới hạn: ', tritoihan)

p0 = 0.05
f = 0.03
n = 800

trikiemdinh = ((f - p0) / np.sqrt((p0*(1-p0))/n)).round(4)
print('- Trị kiểm định: ', trikiemdinh)

print('- p_value: ', stats.norm.cdf(trikiemdinh).round(4))
```
